```python
import pandas as pd

COLUMNS = [
    'option_id',
    'diameter',
    'material',
    'darcy_friction_factor-new_pipe',
    'darcy_friction_factor-decay_rate-min',
    'darcy_friction_factor-decay_rate-max',
    'lifetime-min',
    'lifetime-max',
]

MATERIALS = ['PVC', 'Welded steel']

DIAMS = dict(zip(MATERIALS, [
    list(range(200, 1001, 200)), # for PVC
    [800, 1000, 1250, 1500, 2000] # steel
]))

DARCY_NEW = dict(zip(MATERIALS, [
    0.0015,
    0.045
]))

DARCY_MIN = dict(zip(MATERIALS, [
    0.0,
    0.000259417936211881
]))

DARCY_MAX = dict(zip(MATERIALS, [
    7e-05,
    0.00252348077248057
]))

LIFETIME_MIN = dict(zip(MATERIALS, [
    60,
    90
]))

LIFETIME_MAX = dict(zip(MATERIALS, [
    100,
    130
]))


def option_id_gen(x: int) -> str:
    return f"PI{x:02d}"

pipe_options_df = pd.DataFrame(columns=COLUMNS)

for material in MATERIALS:
    for diam_option in DIAMS[material]:
        pipe_options_df.loc[len(pipe_options_df)] = [
            option_id_gen(len(pipe_options_df) + 1),
            diam_option,
            material,
            DARCY_NEW[material],
            DARCY_MIN[material],
            DARCY_MAX[material],
            LIFETIME_MIN[material],
            LIFETIME_MAX[material]
        ]
```

```python
from water_futures_battle.core.base_model import StaticProperties

pipe_options_sps = StaticProperties('pipe_options-static_properties')
pipe_options_sps['options'] = pipe_options_df.copy()

pipe_options_sps.dump(path='..')
```
