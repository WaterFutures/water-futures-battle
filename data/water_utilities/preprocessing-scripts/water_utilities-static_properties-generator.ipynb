```python
import pandas as pd

wu_pvs = [
    ["PV20"],
    ["PV21"],
    ["PV22"],
    ["PV23"],
    ["PV24", "PV25"],
    ["PV26"],
    ["PV27"],
    ["PV28", "PV29"],
    ["PV30"],
    ["PV31"]
]
wu_pvs = [';'.join(pvs) for pvs in wu_pvs]

wu_df = pd.DataFrame(
    data=wu_pvs,
    columns=['assigned_provinces']
)

def wu_id_gen(x: int) -> str:
    return f"WU{x:02d}"

wu_df['water_utility_id'] = [wu_id_gen(i) for i in range(1, len(wu_df)+1)]
wu_df = wu_df.set_index('water_utility_id').reset_index()
```

```python
from water_futures_battle.core.base_model import StaticProperties

wu_sps = StaticProperties('water_utilities-static_properties')
wu_sps['entities'] = wu_df

wu_sps.dump(path='..')
```
