## Introduction

This supplementary material presents the participants' results in the Battle of the Water Futures and a visual re-interpretation of the takeaway survey responses.

This computational notebook generates the figures of Supplementary Material S4.
The interested reader can explore the original interactive versions of these figures by opening this in Google Colab, where the figures are pre-computed and embedded, so no code needs to be run.

[![Open In Colab](https://colab.research.google.com/assets/colab-badge.svg)](https://colab.research.google.com/github/WaterFutures/water-futures-battle/blob/main/bwf_competition_analysis/supplementary_information.ipynb)

Supplementary Material 3 reports the participants' survey responses in tabular form.

The participating teams' raw results are not publicly available, but they can be reproduced by running the competition again on the anonymised masterplans.

```python
# Imports and helper functions
from pathlib import Path

import data_handling
import plotting_functions

from IPython.display import HTML, display

RENDER_HTML = True
SAVE_SVG = False
if SAVE_SVG:
    Path("2026-08-CPWater-Zanutto_et_al-ST4-figures").mkdir(parents=True, exist_ok=True)
def show(name:str, fig):
    if RENDER_HTML:
        display(HTML(fig.to_html(full_html=False, include_plotlyjs="cdn")))
    else:
        fig.show()
    if SAVE_SVG:
        fig.write_image(f"2026-08-CPWater-Zanutto_et_al-ST4-figures/{name}.svg")

team_data_dict = data_handling.upload_teams_data(teams_dir=Path("..", "teams"))
bwf_statistics = data_handling.get_competition_statistics(team_data_dict)
```

**Figure S15.** Scatter plot of the competing teams on the four-dimensional objective space in each competition stage.
In each panel (competition stage), reliability perception (population-weighted worst reliability across a stage) is plotted against the debt burden (average net present value of debt per capita).
Marker colour indicates affordability distress, from more affordable (blue, low percentage of income spent on water) to less affordable (red, high percentage/distress).
Marker size indicates the solution's GHG emissions in that stage, with larger markers corresponding to higher emissions.
Marker symbol identifies the team.
A solution that dominates all others in a panel would sit in the top-left corner, with the smallest, most blue marker.
