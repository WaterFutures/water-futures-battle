```python
for team, team_data in team_data_dict.items():

    if not team_data.has_submitted_survey:
        continue

    print("---")
    print(team)

    print("Takeaway #1:")
    print(team_data.survey_response.takeaway_1)
    print()

    print("Takeaway #2:")
    print(team_data.survey_response.takeaway_2)
    print()

    print("Takeaway #3:")
    print(team_data.survey_response.takeaway_3)
    print()
```

```python
for team, team_data in team_data_dict.items():

    if not team_data.has_submitted_survey:
        continue

    print("---")
    print(team)
    print(team_data.survey_response.comment)
    print()
```

```python
for team, team_data in team_data_dict.items():

    if not team_data.has_submitted_survey:
        continue

    print("---")
    print()

    print("Ranked feature interest:")
    [print(f"{i}. {f}") for i, f in enumerate(team_data.survey_response.evalutor_preference.split(';')[:-1], start=1)]
    print()

    print("Main barrier using the evaluator:")
    print(team_data.survey_response.evaluator_main_challenge)
    print()

    print("Suggestion:")
    print(team_data.survey_response.evaluator_suggestions)
    print()
```
